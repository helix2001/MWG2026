```python
import pandas as pd

dataset = pd.read_csv(r'e:\MWG_data\Canada_TBCRM_Data_Aug7_2025.csv', sep= ',')
```

```python
print(dataset.columns)
```

```python
flight_list = dataset.drop(axis = 1, labels=['f073_timeof_rx_pos','f080_target_addr', 'lat','lon','f145_fl',
    'f160_gndvec_trk_angle','f160_gndvec_gnd_speed','f170_target_id', 'f210_mops_ver', 'f010_sac', 'f010_sic'],
    inplace=False)

flight_list.drop_duplicates(inplace=True, ignore_index=True)

print(flight_list)
```

```python
reports = dataset.drop(axis = 1, labels=['ots', 'fix_timestamps', 'fix_lats', 'fix_lons',
           'rad_nav_equipment', 'ssr_equipment', 'otherinfo_text',
           'aircraft_type_text'],   inplace=False)

reports.drop_duplicates(inplace=True, ignore_index=True)

print(reports)
```

```python
flight_list.to_csv('flights.split.csv',sep=',')
reports.to_csv('reports.split.csv',sep=',')
```
